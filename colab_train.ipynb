```
# ── Cell 10: Plot reward components over training ─────────────────────────────
import json, glob, torch
from pathlib import Path
import matplotlib.pyplot as plt

# Load reward history from the latest checkpoint
checkpoints = sorted(Path(CHECKPOINT_DIR).glob('checkpoint_step_*.pt'))
if not checkpoints:
    print('No checkpoints found yet.')
else:
    data = torch.load(checkpoints[-1], map_location='cpu', weights_only=False)
    history = data.get('reward_history', [])
    if not history:
        print('Reward history empty.')
    else:
        steps   = [h['step'] for h in history]
        rewards = [h['reward'] for h in history]
        rp = [h['components'].get('retrieval_presence', 0) for h in history]
        te = [h['components'].get('token_efficiency', 0) for h in history]
        lc = [h['components'].get('logical_consistency', 0) for h in history]
        gr = [h['components'].get('grounding', 0) for h in history]

        fig, axs = plt.subplots(2, 1, figsize=(12, 8))
        axs[0].plot(steps, rewards, label='Composite reward', color='black', linewidth=2)
        axs[0].set_ylabel('Reward'); axs[0].set_title('PPO Training — Composite Reward'); axs[0].legend()

        axs[1].plot(steps, rp, label='Retrieval presence')
        axs[1].plot(steps, te, label='Token efficiency')
        axs[1].plot(steps, lc, label='Logical consistency')
        axs[1].plot(steps, gr, label='Grounding')
        axs[1].set_xlabel('Step'); axs[1].set_ylabel('Component score')
        axs[1].set_title('Four Reward Components'); axs[1].legend()

        plt.tight_layout()
        plt.savefig(f'{DRIVE_DIR}/reward_curve.png', dpi=150)
        plt.show()
        print(f'Steps trained: {len(history)}')
        print(f'Final reward: {rewards[-1]:.4f}')
```

```
# ── Cell 11: Eval after training ─────────────────────────────────────────────
!python eval.py --out results_posttrain.json

import json
pre  = json.load(open('results_pretrain.json'))['summary']
post = json.load(open('results_posttrain.json'))['summary']

print('\n── Evaluation Summary ──────────────────────────────')
for system in ['baseline', 'pipeline']:
    if system in pre and system in post:
        print(f'  {system:12}  PRE  EM={pre[system]["em"]:.4f}  F1={pre[system]["f1"]:.4f}')
        print(f'  {system:12}  POST EM={post[system]["em"]:.4f}  F1={post[system]["f1"]:.4f}')
print('────────────────────────────────────────────────────')

# Copy results to Drive for persistence
import shutil
for f in ['results_pretrain.json', 'results_posttrain.json']:
    shutil.copy(f, f'{DRIVE_DIR}/{f}')
print(f'Results saved to {DRIVE_DIR}/')
```
